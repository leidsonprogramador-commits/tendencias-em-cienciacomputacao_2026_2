```python
import math

class Calculadora:
    """
    Uma classe que oferece métodos estáticos para operações matemáticas básicas.
    Cada método contém asserções para verificar a correção do cálculo.
    """

    @staticmethod
    def soma(a: float, b: float) -> float:
        """
        Realiza a soma de dois números.
        Args:
            a (float): O primeiro número.
            b (float): O segundo número.
        Returns:
            float: A soma de a e b.
        """
        return a + b

    @staticmethod
    def subtracao(a: float, b: float) -> float:
        """
        Realiza a subtração de dois números.
        Args:
            a (float): O primeiro número (minuendo).
            b (float): O segundo número (subtraendo).
        Returns:
            float: A diferença entre a e b.
        """
        return a - b

    @staticmethod
    def multiplicacao(a: float, b: float) -> float:
        """
        Realiza a multiplicação de dois números.
        Args:
            a (float): O primeiro número.
            b (float): O segundo número.
        Returns:
            float: O produto de a e b.
        """
        return a * b

    @staticmethod
    def divisao(a: float, b: float) -> float:
        """
        Realiza a divisão de dois números.
        Args:
            a (float): O numerador.
            b (float): O denominador.
        Returns:
            float: O quociente de a por b.
        Raises:
            ValueError: Se o denominador for zero.
        """
        if b == 0:
            raise ValueError("Não é possível dividir por zero.")
        return a / b

    @staticmethod
    def potencia(a: float, b: float) -> float:
        """
        Calcula a potência de um número.
        Args:
            a (float): A base.
            b (float): O expoente.
        Returns:
            float: a elevado à potência b.
        """
        return a ** b

    @staticmethod
    def raiz_quadrada(a: float) -> float:
        """
        Calcula a raiz quadrada de um número.
        Args:
            a (float): O número para o qual a raiz quadrada será calculada.
        Returns:
            float: A raiz quadrada de a.
        Raises:
            ValueError: Se o número for negativo.
        """
        if a < 0:
            raise ValueError("Não é possível calcular a raiz quadrada de um número negativo.")
        return math.sqrt(a)
```

Esta célula define a classe `Calculadora` com métodos estáticos para as operações básicas e tratamento de erros para divisão por zero e raiz quadrada de números negativos.

```python
def obter_numero(prompt: str) -> float:
    """
    Solicita ao usuário que digite um número e valida a entrada.
    Continua solicitando até que um número válido seja fornecido.
    Args:
        prompt (str): A mensagem a ser exibida ao solicitar o número.
    Returns:
        float: O número válido inserido pelo usuário.
    """
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

def validar_resultado(operacao: str, *args: float) -> float:
    """
    Executa uma operação da calculadora e verifica o resultado usando asserções.
    As asserções garantem que o resultado da operação seja o esperado.
    Args:
        operacao (str): O código da operação ('1'-'6').
        *args (float): Os números de entrada para a operação. Pode ser um ou dois.
    Returns:
        float: O resultado da operação.
    Raises:
        ValueError: Se a operação for inválida.
        AssertionError: Se o resultado calculado não corresponder ao esperado.
    """
    a = args[0]
    b = None
    if len(args) > 1:
        b = args[1]

    if operacao == "1": # Soma
        resultado = Calculadora.soma(a, b)
        assert resultado == a + b, f"Erro na soma: esperado {a + b}, obtido {resultado}"
    elif operacao == "2": # Subtração
        resultado = Calculadora.subtracao(a, b)
        assert resultado == a - b, f"Erro na subtração: esperado {a - b}, obtido {resultado}"
    elif operacao == "3": # Multiplicação
        resultado = Calculadora.multiplicacao(a, b)
        assert resultado == a * b, f"Erro na multiplicação: esperado {a * b}, obtido {resultado}"
    elif operacao == "4": # Divisão
        resultado = Calculadora.divisao(a, b)
        # Usar uma tolerância para comparação de floats
        assert abs(resultado * b - a) < 1e-9, f"Erro na divisão: esperado {a / b}, obtido {resultado}"
    elif operacao == "5": # Potência
        resultado = Calculadora.potencia(a, b)
        assert resultado == pow(a, b), f"Erro na potência: esperado {pow(a, b)}, obtido {resultado}"
    elif operacao == "6": # Raiz Quadrada
        resultado = Calculadora.raiz_quadrada(a)
        assert abs(resultado**2 - a) < 1e-9, f"Erro na raiz quadrada: esperado {math.sqrt(a)}, obtido {resultado}"
    else:
        raise ValueError("Operação inválida.")

    return resultado
```

Esta célula contém duas funções auxiliares:

*   `obter_numero(prompt)`: Garante que o usuário insira um número válido.
*   `validar_resultado(operacao, *args)`: Executa a operação e, crucialmente, utiliza `assert` para verificar se o resultado obtido corresponde ao esperado, garantindo a correção dos cálculos. Caso haja uma falha, uma mensagem de erro é exibida.

```python
def menu():
    """
    Exibe um menu interativo para a calculadora, solicita a operação e os números,
    e mostra o resultado. Lida com erros de entrada e de validação de cálculo.
    """
    while True:
        print("\n=== CALCULADORA PYTHON ===")
        print("1 - Soma")
        print("2 - Subtração")
        print("3 - Multiplicação")
        print("4 - Divisão")
        print("5 - Potência")
        print("6 - Raiz Quadrada")
        print("0 - Sair")

        opcao = input("Escolha uma opção: ")

        try:
            if opcao == "0":
                print("Encerrando...")
                break

            if opcao == "6": # Raiz Quadrada exige apenas um número
                a = obter_numero("Digite o número: ")
                resultado = validar_resultado(opcao, a)
            elif opcao in ["1", "2", "3", "4", "5"]:
                a = obter_numero("Primeiro número: ")
                b = obter_numero("Segundo número: ")
                resultado = validar_resultado(opcao, a, b)
            else:
                print("Opção inválida. Por favor, escolha uma opção de 0 a 6.")
                continue

            print(f"Resultado: {resultado}")

        except ValueError as e:
            print(f"Erro: {e}")
        except AssertionError as e:
            print(f"Falha na validação do cálculo: {e}")
        except Exception as e:
            print(f"Ocorreu um erro inesperado: {e}")

if __name__ == "__main__":
    menu()
```

Esta célula implementa a interface do usuário com um menu interativo. Ele solicita a operação desejada e os números, chama a função de validação de resultados e exibe o resultado ou qualquer erro que possa ocorrer (como entrada inválida, divisão por zero, ou uma falha na asserção de cálculo).
